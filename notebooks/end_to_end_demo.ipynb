# ECM Digital Thread Toolkit — End-to-End Demo

Demonstrates the Digital Thread + NLP + KPI/Bottleneck analytics pipeline on the synthetic demonstration dataset (see the README, section "Dataset"). Run `pip install -r ../requirements.txt` first if needed.

````python
import sys
sys.path.insert(0, '../src')

from ecm_digital_thread.data_generation import generate_dataset
from ecm_digital_thread.digital_thread import compute_full_linkage, traceability_coverage_index
from ecm_digital_thread.nlp_pipeline import train_classifier
from ecm_digital_thread.kpi_analytics import kpi_summary, bottleneck_severity_score

df = generate_dataset()
df.head()
````

## Digital Thread integration + Traceability Coverage Index (Eq. 1)

````python
df = compute_full_linkage(df)
print('Overall TCI:', traceability_coverage_index(df), '%')
traceability_coverage_index(df, group_col='month')
````

## NLP classification (TF-IDF + Linear SVM)

````python
result = train_classifier(df)
import pandas as pd
pd.DataFrame(result.report_dict).T[['precision', 'recall', 'f1-score']]
````

## KPI summary + Bottleneck Severity Score (Eq. 3)

````python
monthly = kpi_summary(df, group_col='month')
monthly
````

````python
bss = bottleneck_severity_score(df, group_col='month')
bss[bss['is_bottleneck']]
````
